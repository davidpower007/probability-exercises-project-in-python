```python
import numpy as np
import pandas as pd
#Ejercicio 1

#la probabilidad estimada de que la puntuación sea mayor a 7 o un número par
#Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
n = 1000

dado1 = np.random.randint(1 , 7, size = n)

dado2 = np.random.randint(1, 7, size = n)

suma = dado1 + dado2

probabilidad_7_par = np.mean((suma > 7)  | (suma % 2 ==0))

probabilidad_7_par


#Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
```

```python
#Lleva una cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
n = 1000
contador = 0

dado1 = np.random.randint(1 , 7, size = n)

dado2 = np.random.randint(1, 7, size = n)

suma = dado1 + dado2

for suma in range(n):
    if ((suma > 7)  | (suma % 2 ==0)):
          contador += 1

contador/n
```

```python
#Exercise 2

#Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. . Si tomamos 5 bolas de la caja reemplazándolas. la  probabilidad  de Tomar 3 blancas y 2 rojas

ball_box = ["blancas"]*10 + ["rojas"]*20 + ["verdes"]*30

n = 1000
extracciones = np.random.choice(ball_box, size = (n, 5), replace = True)
#Tomar 3 blancas y 2 rojas
conteo_blancas = np.sum(extracciones == "blancas", axis=1)
conteo_rojas = np.sum(extracciones == "rojas", axis=1)

Tomar_3_blancas_2_rojas = (conteo_blancas == 3) & (conteo_rojas == 2)

probabilidad1 = np.mean(Tomar_3_blancas_2_rojas)
probabilidad1
```

```python
#Todas todas del mismo color.
conteo_blancas = np.sum(extracciones == "blancas", axis=1)
conteo_rojas = np.sum(extracciones == "rojas", axis=1)
conteo_verdes = np.sum(extracciones == "verdes", axis=1)

tomar_mismo_color = (conteo_blancas == 5) | (conteo_rojas == 5) | (conteo_verdes == 5)

probabilidad2 = np.mean(tomar_mismo_color)
probabilidad2
```
